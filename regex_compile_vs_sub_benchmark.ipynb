## **Experiment:** Benchmark comparing the performance of re.compile().sub() versus direct re.sub() for repeated regex usage.

```python
import re
import random
import string
import time



def random_word():
    return "".join(random.choices(string.ascii_letters + "   \t\n", k=20))

words = [random_word() for _ in range(int(1e6))]


# Case 1: re.compile


pattern = re.compile(r"\s+")

start = time.time()
for w in words:
  pattern.sub(" ", w)
compile_time = time.time() - start


# Case 2: re.sub directly


start = time.time()
for w in words:
  re.sub(r"\s+", " ", w)
direct_time = time.time() - start


# Results


print(f"re.compile + sub : {compile_time:.3f} seconds")
print(f"re.sub directly  : {direct_time:.3f} seconds")
print(f"Speedup          : {direct_time / compile_time:.2f}×")
```
